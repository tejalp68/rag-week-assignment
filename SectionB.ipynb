```python
def fixed_chunking(text, chunk_size=50, overlap=10):
    words = text.split()
    chunks = []
    start = 0
    while start < len(words):
        end = start + chunk_size
        chunk = " ".join(words[start:end])
        chunks.append(chunk)
        start += chunk_size - overlap
    return chunks

all_chunks = []
for para in corpus:
    all_chunks.extend(fixed_chunking(para, chunk_size=15, overlap=5))

for i, c in enumerate(all_chunks):
    print(f"Chunk {i}: {c}\n")
```
